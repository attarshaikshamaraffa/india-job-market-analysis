```python
import pandas as pd

df = pd.read_csv("../data/raw/jobs.csv")
print(df.shape)
print(df.columns.tolist())
df.head(10)
```

```python
df.info()
```

```python
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Job_IDs:", df["Job_ID"].duplicated().sum())
```

```python
df.columns = df.columns.str.lower()

text_cols = df.select_dtypes(exclude="number").columns
for c in text_cols:
    df[c] = df[c].str.strip()

df = df.drop_duplicates(subset="job_id")
df["date_posted"] = pd.to_datetime(df["date_posted"], errors="coerce")
df["experience_group"] = df["experience_level"].str.split(" ").str[0]
df.shape
```

```python
df[["salary_lpa", "openings", "applicants", "company_rating"]].describe()
```

```python
skills = df[["job_id", "skills_required"]].copy()
skills["skill"] = skills["skills_required"].str.split(",")
skills = skills.explode("skill")
skills["skill"] = skills["skill"].str.strip()
skills = skills.drop(columns="skills_required")

skills["skill"].value_counts().head(15)
```

```python
print("Bad dates:", df["date_posted"].isnull().sum())
print("Date range:", df["date_posted"].min(), "to", df["date_posted"].max())

df.to_csv("../data/clean/jobs_clean.csv", index=False)
skills.to_csv("../data/clean/job_skills.csv", index=False)
```

```python
import sqlite3

conn = sqlite3.connect("../data/jobs.db")
df.to_sql("jobs", conn, if_exists="replace", index=False)
skills.to_sql("skills", conn, if_exists="replace", index=False)

pd.read_sql("SELECT COUNT(*) AS total_jobs FROM jobs", conn)
```

```python
q1 = """
SELECT experience_group,
       COUNT(*) AS jobs,
       ROUND(AVG(salary_lpa), 1) AS avg_salary_lpa
FROM jobs
GROUP BY experience_group
ORDER BY avg_salary_lpa;
"""
pd.read_sql(q1, conn)
```
